```python
class Account:
    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.balance = balance

    def show_details(self):
        print(f"Account Number: {self.account_number}")
        print(f"Customer Name: {self.customer_name}")
        print(f"Balance: Rs. {self.balance:.0f}")

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit Failed: Amount must be greater than 0")
            return
        self.balance += amount
        print(f"Deposit Successful. New Balance: Rs. {self.balance:.0f}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal Failed: Amount must be greater than 0")
            return
        if amount > self.balance:
            print("Withdrawal Failed: Insufficient balance")
            return
        self.balance -= amount
        print(f"Withdrawal Successful. New Balance: Rs. {self.balance:.0f}")

    def calculate_profit(self):
        # Base version — overridden by children
        return 0


class SavingsAccount(Account):
    def calculate_profit(self):
        profit = self.balance * 0.05
        print(f"Savings Profit (5%): Rs. {profit:.0f}")
        return profit


class CurrentAccount(Account):
    def calculate_profit(self):
        print("Current Account: No profit (transaction-only account)")
        return 0


class BusinessAccount(Account):
    def calculate_profit(self):
        profit = self.balance * 0.07
        print(f"Business Profit (7%): Rs. {profit:.0f}")
        return profit


# ---- Storage for all accounts ----
accounts = {}


def create_account():
    print()
    print("===== CREATE ACCOUNT =====")
    print("1. Savings Account")
    print("2. Current Account")
    print("3. Business Account")

    choice = input("Enter choice (1-3): ").strip()

    if choice not in ("1", "2", "3"):
        print("Invalid choice.")
        return

    acc_num  = input("Enter account number: ")
    cust_name = input("Enter customer name: ")

    try:
        balance = float(input("Enter initial balance: "))
    except ValueError:
        print("Invalid balance.")
        return

    if acc_num in accounts:
        print(f"Account {acc_num} already exists.")
        return

    if choice == "1":
        accounts[acc_num] = SavingsAccount(acc_num, cust_name, balance)
        print(f"Savings Account {acc_num} created.")
    elif choice == "2":
        accounts[acc_num] = CurrentAccount(acc_num, cust_name, balance)
        print(f"Current Account {acc_num} created.")
    elif choice == "3":
        accounts[acc_num] = BusinessAccount(acc_num, cust_name, balance)
        print(f"Business Account {acc_num} created.")


def get_account():
    acc_num = input("Enter account number: ").strip()
    if acc_num not in accounts:
        print(f"Account {acc_num} not found.")
        return None
    return accounts[acc_num]


def do_check_details():
    acc = get_account()
    if acc:
        print()
        acc.show_details()


def do_deposit():
    acc = get_account()
    if not acc:
        return
    try:
        amount = float(input("Enter deposit amount: "))
    except ValueError:
        print("Invalid amount.")
        return
    acc.deposit(amount)


def do_withdraw():
    acc = get_account()
    if not acc:
        return
    try:
        amount = float(input("Enter withdrawal amount: "))
    except ValueError:
        print("Invalid amount.")
        return
    acc.withdraw(amount)


def do_calculate_profit():
    acc = get_account()
    if not acc:
        return
    print()
    acc.calculate_profit()


def main():
    print("=========================================")
    print("     PAKISTANI BANK — ACCOUNT SYSTEM     ")
    print("=========================================")

    while True:
        print()
        print("===== MENU =====")
        print("1. Create Account")
        print("2. Check Account Details")
        print("3. Deposit")
        print("4. Withdraw")
        print("5. Calculate Profit")
        print("6. Exit")

        choice = input("Enter choice (1-6): ").strip()

        if choice == "1":
            create_account()
        elif choice == "2":
            do_check_details()
        elif choice == "3":
            do_deposit()
        elif choice == "4":
            do_withdraw()
        elif choice == "5":
            do_calculate_profit()
        elif choice == "6":
            print("Thank you for using Pakistani Bank. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1-6.")


main()
```
